### 1.4 Look into the loading warning
The warning said that columns 54 and 55 have *mixed types*. Here we find out which columns they are and what is mixed inside them.

```python
mixed_cols = [df.columns[54], df.columns[55]]
print("Columns 54 and 55 are:", mixed_cols)

for c in mixed_cols:
    print(f"\nPython types inside {c}:")
    print(df[c].map(type).value_counts().to_string())

print("\nRISK_CUSTOMER values exactly as stored (repr shows hidden spaces):")
print(df['RISK_CUSTOMER'].map(repr).value_counts().head(10).to_string())
```

**What this shows:** pandas reads a big file in blocks of rows and decides the type block by block. In most blocks `ACID` looked like a number, so pandas stored it as a number, and a number cannot keep leading zeros: `001402238019002` became `1402238019002`. One block contained a value made only of spaces, so pandas kept that whole block as text, zeros included. The same account therefore appears under two different ids. `RISK_CUSTOMER` has the same mix, plus extra spaces (`'2   '` vs `2`).

**length of the account ID may depend on the length of the branch code

**After the fix:** every `ACID` should be text with 15 characters. One row has no `ACID` at all (it held only spaces); it shows as a missing value (`float`) here and is removed in step 1.5.

```python
# Fix the two columns pandas warned about positions 54 and 55
print("Columns 54 and 55:", df.columns[[54, 55]].tolist())

# each column holds a mix of numbers and text
for c in ['ACID', 'RISK_CUSTOMER']:
    print(f"{c} types as first loaded:", df[c].map(type).value_counts().to_dict())
acid_before = df['ACID'].nunique()

# The fix: read both columns again as text, so values are exactly as written in the file (leading zeros kept)
fix = pd.read_csv(file_path, usecols=['ACID', 'RISK_CUSTOMER'], dtype=str)
assert len(fix) == len(df)                                     
df['ACID'] = fix['ACID'].str.strip().replace('', np.nan)
df['RISK_CUSTOMER'] = pd.to_numeric(fix['RISK_CUSTOMER'].str.strip(), errors='coerce').astype('Int64')
del fix

# The check: one account = one id, every id 15 digits, starting with its branch code
branch_code = df['SOL_ID'].astype(str).str.zfill(4)
print(f"\nDistinct accounts: {acid_before:,} before -> {df['ACID'].nunique():,} after")
print("ACID lengths after:", df['ACID'].dropna().str.len().value_counts().to_dict(),
      f"| missing: {df['ACID'].isna().sum()}")
print(f"ACID starts with its 4-digit branch code: {(df['ACID'].str[:4] == branch_code).mean():.1%}")
print("RISK_CUSTOMER after:", df['RISK_CUSTOMER'].value_counts(dropna=False).to_dict())

# Remove extra spaces from every other text column too
text_cols = [c for c in df.columns if not pd.api.types.is_numeric_dtype(df[c]) and c != 'ACID']
for c in text_cols:
    df[c] = df[c].str.strip().replace('', np.nan)
```

### 1.7 Dates
Turn the date columns from text into real dates, then check that every value parsed and that the ranges make sense.

```python
DATE_COLS = ['TREND_DATE', 'OVERDRAFT_LIMIT_EXPIRY_DATE', 'POD_LIMIT_EXPIRY_DATE', 'TOD_LIMIT_EXPIRY_DATE',
             'DATE_OF_BIRTH', 'ACCT_OPN_DATE', 'EXCESS_START_DATE', 'LAST_DR_INT_CYCLE_DATE', 'NEXT_DR_INT_CYCLE_DATE',
             'LAST_CR_INT_CYCLE_DATE', 'NEXT_CR_INT_CYCLE_DATE', 'LAST_CHANGE_TIME']


def to_date(s):
    # parse each distinct text value once, then map back 
    values = list(s.dropna().unique())
    if not values:
        return pd.to_datetime(s)
    parsed = pd.to_datetime(pd.Series(values), errors='coerce', format='mixed')
    return s.map(pd.Series(parsed.values, index=values))


rows = []
for c in DATE_COLS:
    before = int(df[c].notna().sum())
    df[c] = to_date(df[c])
    rows.append([c, before, int(df[c].notna().sum()), df[c].min(), df[c].max()])
date_check = pd.DataFrame(rows, columns=['column', 'values before', 'parsed as date', 'earliest', 'latest'])
if (date_check['values before'] != date_check['parsed as date']).any():
    print('WARNING: some dates did not parse. Check their format before going on.')
date_check
```

### 1.10 How the balance columns fit together
The sample rows suggest two rules used by the bank's system:
- **available balance = cleared balance - reserved balance + limit**, so the limit the customer can really use is `available - cleared + reserved`
- **arrears amount = the part of the available balance below zero**

If both rules hold for most rows, we can trust that "real limit" more than the three limit columns, which do not add up.

```python
real_limit = df['ACCOUNT_AVAILABLE_BALANCE'] - df['ACCOUNT_CLEARED_BALANCE'] + df['ACCOUNT_RESERVED_BALANCE']
arrears_rule = (-df['ACCOUNT_AVAILABLE_BALANCE']).clip(lower=0)

checks = pd.Series({
    'ARR_AMT == part of available balance below zero': np.isclose(df['ARR_AMT'], arrears_rule, atol=1).mean(),
    'real limit == |OVERDRAFT_LIMIT|': np.isclose(real_limit, df['OVERDRAFT_LIMIT'].abs(), atol=1).mean(),
    'real limit == |TOD_LIMIT| + |POD_LIMIT|': np.isclose(real_limit, df['TOD_LIMIT'].abs() + df['POD_LIMIT'].abs(), atol=1).mean(),
    'real limit == |POD_LIMIT|': np.isclose(real_limit, df['POD_LIMIT'].abs(), atol=1).mean(),
}, name='rows where it holds').map('{:.1%}'.format)
print(checks.to_string())

has_some_limit = (df[['OVERDRAFT_LIMIT', 'TOD_LIMIT', 'POD_LIMIT']] != 0).any(axis=1)
print("\nRows with any limit column filled:")
print(f"  real limit above zero   : {(has_some_limit & (real_limit > 1)).sum():,}")
print(f"  real limit zero         : {(has_some_limit & (real_limit.abs() <= 1)).sum():,}  (limit columns filled but nothing usable, e.g. expired)")

print("\nReserved balance on the placeholder rows from 1.9 (if about 10 trillion, the placeholder is a full block on the account):")
print(df.loc[df['ARR_AMT_LKR'] >= SUSPECT, 'ACCOUNT_RESERVED_BALANCE'].describe().apply(lambda v: f'{v:,.2f}').to_string())
```

**Decision:** if the first rule holds for almost every row, Part 3 uses `available - cleared + reserved` as the account's real limit. Otherwise, fall back to `OVERDRAFT_LIMIT`.

## Part 3. Final cleaning


1. `ARR_AMT` / `ARR_AMT_LKR`: on accounts with the bank's block in `ACCOUNT_RESERVED_BALANCE` (99,999,999,999.99 or more, see 1.9 and 1.10) the arrears amount is not real, so it is set to blank. Amounts of 100 billion LKR and above are also set to blank. Only this value is blanked: the rows and all other columns stay.
2. Rows are sorted by `ACID` and `TREND_DATE`, so each account's months are in order.

```python
BLOCK = 99_999_999_999          # the bank's block value in ACCOUNT_RESERVED_BALANCE (see 1.9 and 1.10)
fake = (df['ARR_AMT_LKR'] >= SUSPECT) | (df['ACCOUNT_RESERVED_BALANCE'] >= BLOCK)
df.loc[fake, ['ARR_AMT', 'ARR_AMT_LKR']] = np.nan
print(f"Placeholder arrears amounts set to blank: {int(fake.sum()):,} rows")

df = df.sort_values(['ACID', 'TREND_DATE']).reset_index(drop=True)
del fake
gc.collect()
print(f"Rows: {len(df):,} | columns: {df.shape[1]}")
print(df.columns.tolist())
```

### Overview first: which numeric columns are related?

#### 4.0 Which numeric columns move together?
1 = move together, -1 = move in opposite directions, 0 = no relation. ID columns are left out. Random sample of 400k rows.

```python
NUM = ['NDIA', 'ACCOUNT_STATUS', 'ARR_AMT_LKR', 'OD_EXCESS_LIMIT', 'ACCOUNT_INT_IN_SUSPENSE', 'ACCOUNT_CLEARED_BALANCE',
       'ACCOUNT_AVAILABLE_BALANCE', 'ACCOUNT_RESERVED_BALANCE', 'OVERDRAFT_LIMIT', 'TOD_LIMIT', 'POD_LIMIT',
       'ACCOUNT_INTEREST_RATE', 'INTEREST_RECEIVABLE', 'OCL_CLEARING_TODAY_AMOUNT', 'OCL_RETURNED_CHEQUES_AMOUNT', 'RISK_CUSTOMER']
sample = df[NUM].sample(n=min(400_000, len(df)), random_state=0).astype(float)
corr = sample.corr(method='spearman')
upper = np.triu(np.ones_like(corr, dtype=bool), k=1)
fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(corr, mask=upper, cmap='RdBu_r', vmin=-1, vmax=1, center=0, annot=True, fmt='.2f', annot_kws={'fontsize': 8},
            linewidths=1, linecolor='white', cbar_kws={'shrink': 0.7, 'label': 'Spearman correlation'}, ax=ax)
ax.set_title('Spearman correlation between numeric columns (400k-row sample)')
plt.tight_layout()
plt.show()

pairs = corr.where(np.tril(np.ones_like(corr, dtype=bool), k=-1)).stack().rename('correlation').reset_index()
pairs.columns = ['column A', 'column B', 'correlation']
print('Pairs that move almost together (|correlation| >= 0.8):')
pairs[pairs['correlation'].abs() >= 0.8].sort_values('correlation', key=abs, ascending=False)
```

> **Finding:**
> - `NDIA`, `ACCOUNT_STATUS`, `OD_EXCESS_LIMIT` and `ARR_AMT_LKR` move almost together (0.84-0.99): they all describe the same thing, arrears.
> - `ACCOUNT_INTEREST_RATE` goes up with `NDIA` (0.67) and down with `ACCOUNT_CLEARED_BALANCE` (-0.62), and is -0.90 with `INTEREST_RECEIVABLE`: interest is charged when the account is below zero.
> - `ACCOUNT_AVAILABLE_BALANCE` and `ACCOUNT_CLEARED_BALANCE` are 0.73; `TOD_LIMIT` and `POD_LIMIT` are 0.82 (often given together).
> - `RISK_CUSTOMER`, `OCL_CLEARING_TODAY_AMOUNT` and `OCL_RETURNED_CHEQUES_AMOUNT` are close to 0 with the arrears columns, so they are not related to arrears.
> - `CALCULATED_POD_LIMIT` was an almost exact copy of `OVERDRAFT_LIMIT` (1.00), so it was dropped in Part 2.

### 4D. Arrears: how deep and how big

**What `NDIA` measures (a check, not a chart):** `NDIA` counts the days since the account went **over its limit** (`EXCESS_START_DATE`), and an account is in arrears when it is over its limit (`OD_EXCESS_LIMIT` above 0). For an account with **no** limit, any amount below zero is already over the limit, so it is in arrears straight away. The cell below checks both points on the data.

```python
late = df['NDIA'] > 0
has_start = late & df['EXCESS_START_DATE'].notna()
days_since_start = (df.loc[has_start, 'TREND_DATE'] - df.loc[has_start, 'EXCESS_START_DATE']).dt.days
same = ((days_since_start - df.loc[has_start, 'NDIA']).abs() <= 1).mean()
print(f"NDIA = days from EXCESS_START_DATE to TREND_DATE (within 1 day): {same:.1%} of rows in arrears")
over = df['OD_EXCESS_LIMIT'] > 0
print(f"In arrears when OD_EXCESS_LIMIT > 0: {late[over].mean():.1%} | when OD_EXCESS_LIMIT = 0: {late[~over].mean():.1%}")
del late, has_start, days_since_start, same, over
```

> **Finding:** `NDIA` is the number of days since `EXCESS_START_DATE` in 99.5% of rows in arrears, and 98.6% of rows with `OD_EXCESS_LIMIT` above 0 are in arrears (vs 0.3% of the others). So in this data, **arrears means being over the overdraft limit**, and `NDIA` counts how long the account has been over it.

The last month is left out because it has no next month.

**Look for:** how big "Back to normal" is compared with "Still late".

```python
d = df[['ACID', 'TREND_DATE', 'NDIA']]                         
by_account = d.groupby('ACID')
next_ndia = by_account['NDIA'].shift(-1)                        
next_date = by_account['TREND_DATE'].shift(-1)
has_next_month = next_date == d['TREND_DATE'] + pd.offsets.MonthEnd(1)

late_now = (d['NDIA'] > 0) & (d['TREND_DATE'] < d['TREND_DATE'].max())   # the last month has no "next month"
outcome = np.select([~has_next_month, next_ndia == 0],
                    ['Not in the data next month', 'Back to normal (NDIA = 0)'],
                    default='Still late (NDIA > 0)')
share = (pd.Series(outcome[late_now.to_numpy()]).value_counts(normalize=True)
           .reindex(['Back to normal (NDIA = 0)', 'Still late (NDIA > 0)', 'Not in the data next month']))

fig, ax = plt.subplots(figsize=(9, 3.5))
bars = ax.barh(share.index, share.values, color=[TEAL, ORANGE, GREY], height=0.6)
ax.bar_label(bars, labels=[f'{v:.0%}' for v in share.values], padding=3)
ax.invert_yaxis()
ax.set_xlim(0, 1)
ax.xaxis.set_major_formatter(PCT)
ax.grid(axis='y', visible=False)
start, end = df['TREND_DATE'].min(), df['TREND_DATE'].max()
ax.set_title(f'Accounts late in a month (NDIA > 0): what happens the next month?\n'
             f'All months combined, {start:%b %Y} - {end:%b %Y}')
plt.tight_layout()
plt.show()
del d, by_account, next_ndia, next_date, has_next_month, late_now, outcome
```

> **Finding (all months combined, Jan 2024 - Sep 2026):**
> - Of the accounts late in a month (`NDIA > 0`), the next month **33%** were back to normal, **57%** were still late, and **10%** were no longer in the data.
> - Arrears are "sticky": more than half of late accounts were still late a month later. But about 1 in 3 recovered each month, so many arrears are solved within a few months.
> - About 1 in 10 late accounts disappeared the next month, probably closed or written off.
> - Being late in one month is a strong sign of being late the next month, which is useful for the predictive part.

#### 4.16 Is there a seasonal pattern?
**Why this chart:** chart 4.2 showed a drop every December. This checks the pattern month by month across years.

**How to read it:** each cell = share of accounts with `NDIA > 0`, by year (rows) and calendar month (columns) of `TREND_DATE`.

**Look for:** months that are lighter or darker in every year.

```python
season = (df['NDIA'] > 0).groupby([df['TREND_DATE'].dt.year, df['TREND_DATE'].dt.month]).mean().unstack()
season.columns = [pd.Timestamp(2000, m, 1).strftime('%b') for m in season.columns]
fig, ax = plt.subplots(figsize=(12, 3.2))
sns.heatmap(season, annot=True, fmt='.1%', cmap='Blues', linewidths=2, linecolor='white', annot_kws={'fontsize': 9},
            cbar_kws={'label': 'share with NDIA > 0', 'format': PCT}, ax=ax)
ax.set_title('Share of accounts with NDIA > 0, by year and month')
ax.set_xlabel('month of TREND_DATE')
ax.set_ylabel('year of TREND_DATE')
plt.tight_layout()
plt.show()
```

> **Finding:**
> - **December is the lowest month every year:** 4.9% of accounts in arrears in Dec 2024 and 4.8% in Dec 2025, then back up to about 6.2-6.4% in January. This repeats every year, so something happens at year-end (customers paying before year-end, or a bank clean-up).
> - **2024 was the worst year** (about 7-8.5% of accounts in arrears); 2025 (about 5.6-6.9%) and 2026 (about 5.4-7.2%) are better, so arrears improved after 2024.
> - **One-off months:** Nov 2025 was unusually high (8.1% vs 7.2% in Nov 2024) and Mar 2026 unusually low (5.4%).
> - Because of this pattern, months should be compared with the same month of another year (e.g. December with December), and the month of the year should be considered in the predictive part.
